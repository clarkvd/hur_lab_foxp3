```python
from Bio.Seq import Seq

# Example full sequences
full_sequences = [
    "GNAGATGGCGCTTGAACAGAAAGATTAGGAAAATTTCCACAGTCGACGCGTTGAATTCATCAGCTGGGAGAACAATCATC",
    "GNAGATGGCGCTTGAACAGAAAGATTAGGAAAATTGCACAAATCGAAACAGGAAATTCATCAGCTGGGAGAACAATCATC",
    "GNAGATGGCGCTTGAACAGAAAGATTAGGAAAATTTGTACATTCGAGTACAGGAATTCATCAGCTGGGAGAACAATCATC",
    "GNAGATGGCGCTTGAACAGAAAGATTAGGAAAATTGGGATGCTCGAGCTAGTAAATTCATCAGCTGGGAGAACAATCATC",
    "GNAGATGGCGCTTGAACAGAAAGATTAGGAAAATTGATTCATTCGAAGTCAGTAATTCATCAGCTGGGAGAACAATCATC"
]

# Example trimmed sequences
trimmed_sequences = [
    "TCCACAGTCGACGCGTTG",
    "GCACAAATCGAAACAGGA",
    "TGTACATTCGAGTACAGG",
    "GGGATGCTCGAGCTAGTA",
    "GATTCATTCGAAGTCAGT"
]

# Print reverse complements
for i, (full, trimmed) in enumerate(zip(full_sequences, trimmed_sequences), 1):
    full_rc = str(Seq(full).reverse_complement())
    trimmed_rc = str(Seq(trimmed).reverse_complement())
    print(f"Full sequence #{i}: {full_rc}")
    print(f"Trimmed sequence #{i}: {trimmed_rc}")
```
